```python
import random
import pandas as pd

categories = [
    "Fiction",
    "Education",
    "Horror",
    "Romance",
    "Science",
    "Children"
]

df = pd.read_sql("SELECT book_id FROM books", engine)

with engine.begin() as conn:
    for book_id in df["book_id"]:
        conn.execute(
            text("""
                UPDATE books
                SET category=:category
                WHERE book_id=:book_id
            """),
            {
                "category": random.choice(categories),
                "book_id": int(book_id)
            }
        )

print("Categories added successfully!")
```
